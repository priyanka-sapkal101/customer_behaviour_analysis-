## We fill the miising values in rating with median because mean contain outliers and median is robust and instead of overall median we are going to
## use the median of each category to replace the missing value with thaat particular category.
## fo ex- if we want to fill missing value for clothing we use the median of clothing category and same for all other category.


```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x : x.fillna(x.median()))
```

```python
#Snake Casing - all column names with lowercase and underscore
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ", "_")
df=df.rename(columns={"purchase_amount_(usd)":"purchased_amount"})
```

```python
#create a new column
labels = ['Young','Young_Adult','Middle_Age','Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)
```

```python
frequency_mapping={
    'Fortnightly': 14,
    'Weekly': 7,
    'Bi-Weekly':14,
    'Monthly':30,
    'Quarterly': 90,
    'Annually': 365,
    'Every 3 Months': 90
}
df['purchased_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
    
```
